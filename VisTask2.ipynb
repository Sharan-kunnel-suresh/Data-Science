```python
# Import libraries
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import io
from google.colab import files
```

```python
uploaded = files.upload()
filename = list(uploaded.keys())[0]
data = pd.read_csv(io.BytesIO(uploaded[filename]))
display(data.head())
data.info()
```

```python
# --------------------------
# Step 2: Visualize the data using pyplot
# --------------------------

# 1️⃣ Scatter plot — Age vs Blood Pressure
plt.figure(figsize=(8, 5))
plt.scatter(data['Your Code'], data['Your Code'], c='g', edgecolor='black')
plt.title('Your Code')
plt.xlabel('Your Code')
plt.ylabel('Your Code')
plt.grid(True)
plt.show()
```

```python
# 2️⃣ Histogram — Cholesterol Distribution
plt.figure(figsize=(8, 5))
plt.hist(data['Cholesterol'], bins=15, color='salmon', edgecolor='black')
plt.title('Your Code')
plt.xlabel('Your Code')
plt.ylabel('Your Code')
plt.grid(True)
plt.show()
```

```python
# 3️⃣ Box plot — Heart Rate by Diagnosis
plt.figure(figsize=(8, 5))
data.boxplot(column='Your Code', by='Your Code', grid=False, patch_artist=True)
plt.title('Your Code')
plt.suptitle('')
plt.xlabel('Your Code')
plt.ylabel('Your Code')
plt.show()
```

```python
# --------------------------
# Exercise 1: Pie Chart — Diagnosis Distribution
# --------------------------

diagnosis_counts = data['Your Code'].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(
    diagnosis_counts,
    labels=diagnosis_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['Your Code']
)
plt.title('Your Code')
plt.show()
```

```python
# --------------------------
# Exercise 2: Line Chart — Average Blood Pressure by Age Group
# --------------------------

# Define age bins
bins = [20, 30, 40, 50, 60, 70, 80]
labels = ['Your Code']

data['Your Code'] = pd.cut(data['Your Code'], bins=bins, labels=labels, right=False)

# Calculate mean blood pressure for each group
bp_by_age = data.groupby('Your Code')['Your Code'].mean()

# Plot
plt.figure(figsize=(8, 5))
plt.plot(bp_by_age.index, bp_by_age.values, marker='o', linestyle='-', color='teal')
plt.title('Your Code')
plt.xlabel('Your Code')
plt.ylabel('Your Code')
plt.grid(True)
plt.show()
```

```python
# --------------------------
# Exercise 3: Multi-Plot Dashboard
# --------------------------

fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# Scatter Plot — Age vs Blood Pressure
axs[0, 0].scatter(data['Your Code'], data['Your Code'], c='skyblue', edgecolor='black')
axs[0, 0].set_title('Your Code')
axs[0, 0].set_xlabel('Your Code')
axs[0, 0].set_ylabel('Your Code')

# Histogram — Cholesterol
axs[0, 1].hist(data['Your Code'], bins=15, color='salmon', edgecolor='black')
axs[0, 1].set_title('Your Code')
axs[0, 1].set_xlabel('Your Code')
axs[0, 1].set_ylabel('Your Code')

# Box Plot — Heart Rate by Diagnosis
data.boxplot(column='Your Code', by='Your Code', ax=axs[1, 0], grid=False, patch_artist=True)
axs[1, 0].set_title('Your Code')
axs[1, 0].set_xlabel('Your Code')
axs[1, 0].set_ylabel('Your Code')

# Pie Chart — Diagnosis Distribution
axs[1, 1].pie(
    diagnosis_counts,
    labels=diagnosis_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=['Your Code']
)
axs[1, 1].set_title('Your Code')

plt.suptitle('Dashboard', fontsize=16)
plt.tight_layout()
plt.show()
```
